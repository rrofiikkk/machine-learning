```python
import pandas as pd 
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/rrofiikkk/machine-learning/refs/heads/main/churn-telecom.csv"
df = pd.read_csv(url)
df.info()
```

```python
df.head()
```

```python
df.describe()
```

```python
df['Churn'] = df['Churn'].map({'No': 0, 'Yes': 1})
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Scatter plot
plt.figure(figsize=(10, 6))
scatter = plt.scatter(
    df['tenure'], 
    df['MonthlyCharges'], 
    c=df['Churn'], 
    cmap=plt.cm.get_cmap('coolwarm', 2),  # Gunakan 2 warna saja: 0 dan 1
    alpha=0.7
)

# Titik dan anotasi kotak
plt.text(5, 105, 'N1', fontsize=12, ha='center', color='black', fontweight='bold', backgroundcolor='white')
plt.gca().add_patch(plt.Rectangle((0, 20), 10, 100, linewidth=2, edgecolor='red', facecolor='none'))

# Legend dengan warna eksplisit untuk 0 dan 1
legend_labels = [
    mpatches.Patch(color=plt.cm.coolwarm(0.0), label='No Churn'),   # warna biru
    mpatches.Patch(color=plt.cm.coolwarm(1.0), label='Churn')       # warna merah
]
plt.legend(handles=legend_labels, title='Churn')

# Axis dan title
plt.title('MonthlyCharges vs Tenure', fontsize=14)
plt.xlabel('Tenure', fontsize=12)
plt.ylabel('MonthlyCharges', fontsize=12)
plt.grid(True)
plt.show()
```
