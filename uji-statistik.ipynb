# Import Library

```python
import pandas as pd 
import numpy as np 
import scipy.stats as st 
```

# Read Dataset

**Apakah discount yang diberikan efektif?**

```python
df = pd.read_csv("e-commerce_example_dataset.csv")
df.info()
```

gmv = Jumlah uang yang dikeluarkan oleh customer. <br>
discount = apakah customer diberi diskon atau tidak

# Hypothesis Testing

## T-test

```python
df['discount'].value_counts()
```

## Langkah 1

H0 : rata-rata gmv discount **sama dengan** gmv non-discount <br>
H1 : rata rata gmv discount **berbeda signifikan** dengan gmv non-discount

```python
discount_df = df[df['discount']=='discount']
non_discount_df = df[df['discount']=='not-discount']
```

## Langkah 2

```python
discount_df['gmv'].mean()
```

```python
non_discount_df['gmv'].mean()
```

## Langkah 3

```python
ttest = st.ttest_ind(a=discount_df['gmv'], b=non_discount_df['gmv'])
pvalue = ttest.pvalue
```

statistic -> sebaran t nya. sedangkan p-value-> adalah resikonya. 

```python
print("P-value: ", pvalue)
```

## Langkah 4

```python
if pvalue >= 0.05:
    print("Terima H0 artinya tidak cukup bukti menyatakan bahwa rata-rata gmv yang diberi discount berbeda signifikan")
else:
    print("Terima H1 artinya cukup bukti menyatakan bahwa rata-rata gmv yang diberi discount berbeda signifikan")
```

Pengujian ini dilakukan untuk menjawa "apakah discount ini efektif?"
Menurut statistik gmv yang customer yang diberi discount dengan orang yang tidak diberi discount sudah cukup berbeda signifikan. -> Maka secara bisnis perlu di olah dan analisis kembali.

# ANOVA

**Apakah ada perbedaan signifikan antara masing-masing species untuk panjang sepal length?**

```python
df = pd.read_csv("iris.csv")
df.head()
```

## Langkah 1

H0 : Rata-rata panjang sepal length **sama** untuk masing-masing spesies <br>
H1 : Ada sepasang species yang panjang sepal length **berbeda signifikan** <br>


```python
df['Species'].value_counts()
```

## Langkah 2

```python
setosa = df[df['Species']=='Iris-setosa']
versicolor = df[df['Species']=='Iris-versicolor']
virginica = df[df['Species']=='Iris-virginica']
```

```python
setosa['SepalLengthCm'].mean()
```

```python
versicolor['SepalLengthCm'].mean()
```

```python
virginica['SepalLengthCm'].mean()
```

## Langkah 3

```python
annova_result = st.f_oneway(setosa['SepalLengthCm'], versicolor['SepalLengthCm'], virginica['SepalLengthCm'])
```

```python
pvalue = annova_result.pvalue
print("P-value:", pvalue)
```

## Langkah 4

```python
if pvalue >= 0.05:
    print("Terima H0 artinya tidak cukup bukti menyatakan bahwa sepal length mampu membedakan species")
else:
    print("Terima H1 artinya cukup bukti menyatakan bahwa sepal length mampu membedakan species")
```
